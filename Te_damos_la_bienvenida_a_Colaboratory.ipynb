```
# Primero, descargamos el archivo CSV desde GitHub
!wget https://raw.githubusercontent.com/energy-modelling-toolkit/hydro-power-database/master/data/jrc-hydro-power-plant-database.csv

# Importamos pandas para manejar el archivo CSV
import pandas as pd

# Cargar el archivo CSV
df = pd.read_csv('jrc-hydro-power-plant-database.csv')

# Mostrar toda la base de datos
pd.set_option('display.max_rows', None)  # Esto permite ver todas las filas
pd.set_option('display.max_columns', None)  # Muestra todas las columnas

# Mostrar el contenido de la base de datos
df

```

```
# Ver información básica sobre el DataFrame
df.info()

# Ver estadísticas descriptivas para identificar posibles valores atípicos
df.describe()

```

```
# Comprueba si las columnas existen en el DataFrame antes de eliminarlas
columns_to_drop = ['pypsa_id', 'GEO', 'WRI']
existing_columns = df.columns

# Sólo eliminar columnas que existan en el DataFrame
columns_to_drop = [col for col in columns_to_drop if col in existing_columns]

# Drop the columns
df = df.drop(columns=columns_to_drop)

# Verificar que las columnas han sido eliminadas
df.info()
```

```
# Eliminar filas duplicadas
df = df.drop_duplicates()
```

```
# Rellenar valores faltantes con la media de cada columna numérica
numeric_columns = df.select_dtypes(include=['number']).columns  # Select only numeric columns
df[numeric_columns] = df[numeric_columns].fillna(df[numeric_columns].mean())
```

```
# Convertir 'installed_capacity_MW' a tipo numérico
df['installed_capacity_MW'] = pd.to_numeric(df['installed_capacity_MW'], errors='coerce')

# Convertir 'lat' y 'lon' a tipo flotante
df['lat'] = pd.to_numeric(df['lat'], errors='coerce')
df['lon'] = pd.to_numeric(df['lon'], errors='coerce')
```

```
# Identificar valores atípicos con el método de Z-score
from scipy.stats import zscore

# Calcular el Z-score de las columnas numéricas
z_scores = zscore(df.select_dtypes(include=['float64', 'int64']))

# Filtrar filas que tengan valores atípicos (z-score > 3)
df_clean = df[(z_scores < 3).all(axis=1)]
```

```
# Establecer todas las entradas de 'type' en minúsculas
df['type'] = df['type'].str.lower()

# Reemplazar valores inconsistentes
df['type'] = df['type'].replace({'hydro': 'Hydropower', 'run-of-river': 'Run-of-River'})
```

```
# Guardar el DataFrame limpio en un nuevo archivo CSV
df.to_csv('jrc-hydro-power-plant-database-clean.csv', index=False)
```

```
# Calcular las estadísticas básicas de las variables cuantitativas
# Seleccionamos solo las columnas numéricas
numerical_columns = df.select_dtypes(include=['float64', 'int64'])

# Calcular la media, mediana y desviación estándar
mean_values = numerical_columns.mean()  # Media
median_values = numerical_columns.median()  # Mediana
std_dev_values = numerical_columns.std()  # Desviación estándar

# Mostrar los resultados
print("Media de las columnas numéricas:")
print(mean_values)
print("\nMediana de las columnas numéricas:")
print(median_values)
print("\nDesviación estándar de las columnas numéricas:")
print(std_dev_values)
```
